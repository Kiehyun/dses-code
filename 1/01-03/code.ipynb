**code 36.**

```python
import math

def sinc(x):
    """
    sinc 함수를 계산하는 함수
    sinc(x) = sin(x)/x (x ≠ 0일 때)
    sinc(0) = 1 (x = 0일 때, 극한값)
    """
    try:
        result = math.sin(x) / x
    except ZeroDivisionError:  # x = 0이면 0으로 나누기 오류
        result = 1  # 극한값 sinc(0) = 1
    return result

print("sinc(0) :", sinc(0))  # 예외 처리로 1 반환
print("sinc(0.20) :", sinc(0.20))
```
